### Subfigure 3Sa: Decomposition Performance
With component-based decomposition approaches, we can evaluate the effectiveness of data decomposition through variance-explained. Here, there's two variances we're interested in: (1) the variance explained in the global measurements via RNA-seq and (2) the variance explained in the global measurements, post-deflation, via acetylation. The former can help us to pick a number of components to use in the deflation step while the latter can identify if there are acetylation-specific trends undiscoverable through RNA-seq expression.

```python
from figure3Sa import make_figure as make_figure_3sa
fig_3sa = make_figure_3sa()
```

Four components seems to work for the initial deflation stage; past four, the components capture less than 2% of the variance in global measurements, indicating they may be tuned to noise or rare data trends. We'll keep the same 2% threshold for the acetylation stage, finding that four components is also a good fit for this dataset.

Strikingly, the second-stage acetylation PLSR does seem to capture a large amount of global expression patterns inexplainable by RNA-seq. The four second-stage PLSR components, in total, capture roughly 50% of the variance inexplainable by RNA-seq alone, amounting to around 15% of the variance in the global measurements.
