## Setup

### Imports

```python
import os

import polars as pl
```

### Constants

```python
# Jupyter server should be running from src/python
CACHE_DIR = os.path.join(
    os.path.dirname(os.path.dirname(os.getcwd())),
    "analysis",
    "dylan",
    "_cache",
)
```

## Write keepX CSVs

### Utility function write keepX CSV based off of design CSV

```python
def write_keepx(subset_id, design, n_components, keep_x):
    pl.read_csv(
        os.path.join(
            CACHE_DIR,
            "diablo",
            f"{subset_id}_{design}-design.csv"
        )
    ).select("").with_columns(*[
        pl.lit(keep_x).alias(f"{c}")
        for c in range(n_components)
    ]).write_csv(
        os.path.join(
            CACHE_DIR,
            "diablo",
            f"{subset_id}_{design}-keepX.csv"
        ),
        include_header=False
    )
```

### NPM1 vs. WT

#### Black patients

```python
# null design
write_keepx("B_NPM1-WT", "null", 3, 16)

# PLS-DA design
write_keepx("B_NPM1-WT", "plsda", 2, 8)
```

#### White patients

```python
# null design
write_keepx("W_NPM1-WT", "null", 5, 16)

# PLS-DA design
write_keepx("W_NPM1-WT", "plsda", 7, 8)
```

### NRAS vs. WT

#### Black patients

```python
# null design
write_keepx("B_NRAS-WT", "null", 3, 16)

# PLS-DA design
write_keepx("B_NRAS-WT", "plsda", 2, 16)
```

#### White patients

```python
# null design
write_keepx("W_NRAS-WT", "null", 2, 16)

# PLS-DA design
write_keepx("W_NRAS-WT", "plsda", 4, 16)
```
