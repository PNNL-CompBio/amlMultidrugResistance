### Subfigure S2b: Young Patient Comparisons
While age did not exhibit differential effects across race, it may be useful to compare young Black AML patients to young White AML patients to see if there's differences in how their AML manifests. Here, we compare 'omic measurements between Black and White AML patients between the ages of 15 and 39 (inclusive).

```python
from figureS2b import make_figure as make_figure_s2b
fig_s2b = make_figure_s2b()
```

Indeed, there are some differences in measurements between young Black and White AML patients, though small counts within 'omes indicate it's unlikely that there's larger immunological trends separating young Black and White AML patients. Further examination of the enrichment results, however, does indicate some potential differences in transcriptomic/proteomic expression between young patients.

Starting with the global measurements, we find that young Black AML patients exhibit elevations in signatures related to transcription and splicing pathways across the Gene Ontology libraries. Note that the log-fold change here considers the mean log2 expression of the young White AML patients subtracted from the young Black AML patients, so a positive NES indicates over-expression in Black patients while a negative NES would indicate over-expression in White patients.

```python
import matplotlib.pyplot as plt
from pilot.figures.figure_setup import get_setup

fig, axes = get_setup(
    1,
    3,
    fig_params={
        "figsize": (9, 3),
        "width_ratios": [1, 1.1, 1]
    }
)

global_bio = plt.imread(
    "young_aml/global/GO_Biological_Process_2025_overexpressed.png"
)
global_cell = plt.imread(
    "young_aml/global/GO_Cellular_Component_2025_overexpressed.png"
)
global_molecular = plt.imread(
    "young_aml/global/GO_Molecular_Function_2025_overexpressed.png"
)

axes[0].imshow(global_bio)
axes[1].imshow(global_cell)
axes[2].imshow(global_molecular)

for ax in axes:
    ax.set_axis_off()
```

Conversely, looking into RNA-seq measurements, we observe that young White AML patients exhibit elevations in inflammation-related signatures. Some of this may be explained by the observed elevations in granule-related signatures; granules are specialized components of differentiated granulocytes that (1) can be pro-inflammatory and (2) may indicate further differentiation of GMP-like AML.

```python
import matplotlib.pyplot as plt
from pilot.figures.figure_setup import get_setup

fig, axes = get_setup(
    1,
    3,
    fig_params={
        "figsize": (9, 3),
        "width_ratios": [1, 0.98, 1.03]
    }
)

rna_bio = plt.imread(
    "young_aml/rna/GO_Biological_Process_2025_underexpressed.png"
)
rna_cell = plt.imread(
    "young_aml/rna/GO_Cellular_Component_2025_underexpressed.png"
)
rna_msig = plt.imread(
    "young_aml/rna/MSigDB_Hallmark_2020_underexpressed.png"
)

axes[0].imshow(rna_bio)
axes[1].imshow(rna_msig)
axes[2].imshow(rna_cell)

for ax in axes:
    ax.set_axis_off()
```
