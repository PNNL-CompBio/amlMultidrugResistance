### Subfigure S1c: Mutation waterfall
Given the importance of mutations in AML manifestation, we compare the presence of recurrent mutations across our Black and White AML patients. The below waterfall plots consider only mutations present in at least five White or Black patients, and the mutations in each waterfall are listed in order of frequency for each race.

```python
from figureS1c import make_figure as make_figure_S1c
fig_S1c = make_figure_S1c()
```

There indeed to be some differences in mutation frequencies across race. Some of this is known bias, as White patients were preferentially selected for FLT3-ITD mutations; Black patients, however, seem to exhibit some increases in DNMT3A and NRAS mutations.

### Subfigure S1d: Mutation counts and NPM1/FLT3-ITD justification
Given the clinical importance of NPM1 and FLT3-ITD, as well as their known differences in outcomes across Black and White patients, we've chosen to further examine the effects of these mutations in our multi-omic datasets. Here, we show the number of patients with each mutation across race.

```python
from figureS1d import make_figure as make_figure_S1d
fig_S1d = make_figure_S1d()
```

The later panels in this figure compare the quantities of patients with a single mutation (or the NPM1/FLT3-ITD double mutation) when two of the more frequent mutations--DNMT3A and NRAS--are included. Throughout our analyses, we look to isolate covariates in mutational analyses by examining wild-type patients against those with only the mutation of interest; to evaluate the effect of NPM1, we compare NPM1+/FLT3-ITD- against NPM1-/FLT3-ITD- patients, not all NPM1- patients, to remove potential confounding mutations. Introducing DNMT3A/NRAS further reduces our power in these analyses, and limited single-mutation patients for these groups may limit statistical power.

### Subfigure S1e: Mutation and race co-occurrence
Subfigure 1c indicated some mutations may exhibit differences in frequency across race. To corroborate, we use Fisher's Exact Test to compare mutation and race and observe if there's significant differences in mutation frequencies across race.

```python
from figureS1e import make_figure as make_figure_S1e
fig_S1e = make_figure_S1e()
```

Indeed, we observe some association between race and DNMT3A/FLT3-ITD mutations. FLT3-ITD, as expected, is more prominent in White patients while DNMT3A is more frequently mutated in Black patients.
